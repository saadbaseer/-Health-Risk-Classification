```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)
```

```python
df=pd.read_csv('diabetes.csv')
```

```python
x=df.drop('Outcome',axis=1)
y=df['Outcome']
```

```python
x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

```

```python
scaler = StandardScaler()

x_train_scaled = scaler.fit_transform(x_train)

x_test_scaled = scaler.transform(x_test)
```

```python
x_train_scaled
```

```python
svm_model = SVC(
    kernel="rbf",
    C=1.0,
    gamma="scale",
    random_state=42
)

```

```python
svm_model.fit(x_train_scaled, y_train)
```

```python
y_pred_svm = svm_model.predict(x_test_scaled)
```

```python
y_pred_svm
```

```python
svm_accuracy = accuracy_score(y_test, y_pred_svm)

print("SVM Accuracy:", svm_accuracy)
print("SVM Accuracy (%):", svm_accuracy * 100)
```

```python
print("SVM Classification Report")
print("--------------------------------")

print(
    classification_report(
        y_test,
        y_pred_svm,
        target_names=["No Diabetes", "Diabetes"]
    )
)
```

```python
cm_svm = confusion_matrix(y_test, y_pred_svm)

print("SVM Confusion Matrix:")
print(cm_svm)
```

```python
plt.figure(figsize=(6, 5))

plt.imshow(cm_svm)

plt.title("SVM Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")

plt.xticks([0, 1], ["No Diabetes", "Diabetes"])
plt.yticks([0, 1], ["No Diabetes", "Diabetes"])

for i in range(2):
    for j in range(2):
        plt.text(j, i, cm_svm[i, j], ha="center", va="center")

plt.colorbar()
plt.show()
```

```python
svm_precision = precision_score(y_test, y_pred_svm)

svm_recall = recall_score(y_test, y_pred_svm)

svm_f1 = f1_score(y_test, y_pred_svm)

print("SVM Results")
print("----------------------")
print("Accuracy :", svm_accuracy)
print("Precision:", svm_precision)
print("Recall   :", svm_recall)
print("F1 Score :", svm_f1)
```

# Dicision Tree Classifier

```python
decision_tree = DecisionTreeClassifier(
    criterion="gini",
    max_depth=5,
    random_state=42
)

decision_tree.fit(x_train, y_train)
```

```python
y_pred_tree = decision_tree.predict(x_test)
```

```python
y_pred_tree
```

```python
tree_accuracy = accuracy_score(y_test, y_pred_tree)

print("Decision Tree Accuracy:", tree_accuracy)
print("Decision Tree Accuracy (%):", tree_accuracy * 100)
```

```python
print("Decision Tree Classification Report")
print("--------------------------------------")

print(
    classification_report(
        y_test,
        y_pred_tree,
        target_names=["No Diabetes", "Diabetes"]
    )
)
```

```python
cm_tree = confusion_matrix(y_test, y_pred_tree)

print("Decision Tree Confusion Matrix:")
print(cm_tree)
```

```python
plt.figure(figsize=(6, 5))

plt.imshow(cm_tree)

plt.title("Decision Tree Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")

plt.xticks([0, 1], ["No Diabetes", "Diabetes"])
plt.yticks([0, 1], ["No Diabetes", "Diabetes"])

for i in range(2):
    for j in range(2):
        plt.text(j, i, cm_tree[i, j], ha="center", va="center")

plt.colorbar()
plt.show()
```

```python
tree_precision = precision_score(y_test, y_pred_tree)

tree_recall = recall_score(y_test, y_pred_tree)

tree_f1 = f1_score(y_test, y_pred_tree)

print("Decision Tree Results")
print("----------------------")
print("Accuracy :", tree_accuracy)
print("Precision:", tree_precision)
print("Recall   :", tree_recall)
print("F1 Score :", tree_f1)
```

# Compearin Models

```python
results = pd.DataFrame({
    "Model": ["SVM", "Decision Tree"],
    "Accuracy": [svm_accuracy, tree_accuracy],
    "Precision": [svm_precision, tree_precision],
    "Recall": [svm_recall, tree_recall],
    "F1 Score": [svm_f1, tree_f1]
})

results
```
